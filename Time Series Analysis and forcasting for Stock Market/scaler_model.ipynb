```python
from IPython import get_ipython
from IPython.display import display
# %% [markdown]
# Fit the scaler on training data
# %%
from sklearn.preprocessing import StandardScaler
import numpy as np # Assuming X_train is a numpy array for demonstration
import joblib # Import joblib to save the scaler

#You fit the scaler on your training data to learn the scaling parameters (mean, std, min, max, etc.).
# Define or load X_train before using it
# For example, generating some sample data:
X_train = np.array([[1, 2], [3, 4], [5, 6]])

# Use StandardScaler as originally intended, not MinMaxScaler
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)  # Learn params & transform train data

# Save the scaler object that was just trained on X_train (2 features)
# Make sure the directory exists or change the path
try:
    joblib.dump(scaler, '/content/scaler.save')
    print("Scaler saved successfully.")
except Exception as e:
    print(f"Error saving scaler: {e}")


# %% [markdown]
# Transform test/new data
# %%
# Define or load X_test before using it
# For example, generating some sample data:
X_test = np.array([[7, 8], [9, 10]]) # Replace with your actual test data

# Use the scaler trained above to transform X_test
X_test_scaled = scaler.transform(X_test)

# %% [markdown]
# Load scaler in Colab and use it:
# %%
# Upload scaler.pkl to Colab or load from drive

# Import the joblib module (already imported above)
# import joblib

# Load the scaler object. This should now load the scaler saved in the previous step
# which was trained on 2 features.
try:
    loaded_scaler = joblib.load('/content/scaler.save')
    print("Scaler loaded successfully.")
except FileNotFoundError:
    print("Error: scaler.save not found. Make sure you ran the cell that saves the scaler.")
    loaded_scaler = None # Set to None if loading fails to avoid further errors


# Define or load X_new before using it
# For example, generating some sample new data:
X_new = np.array([[11, 12], [13, 14]]) # Replace with your actual new data

# Only transform if the scaler was loaded successfully
if loaded_scaler is not None:
    # Use the loaded scaler (which should now be StandardScaler trained on 2 features)
    X_new_scaled = loaded_scaler.transform(X_new)
    print("X_new_scaled:", X_new_scaled)
else:
    print("Could not transform X_new as the scaler failed to load.")
```
